# Rock, Paper, Scissors (RPS)

**Rock, Paper, Scissors** is a simple hand game used to make decisions or play for fun.  
It involves two players who simultaneously choose one of three options: `rock`, `paper`, or `scissors`. The winner is determined based on predefined rules.

## How does it work?

- **Rock beats Scissors** (rock crushes scissors)  
- **Scissors beats Paper** (scissors cut paper)  
- **Paper beats Rock** (paper covers rock)  

If both players choose the same option, it’s a **tie**.

## Why is it useful?

- **Conflict resolution**: A quick, simple, and fair way to decide who goes first or gets something.  
- **Simplicity**: No equipment, no other junk—just your hands.  
- **Probability learning**: Teaches randomness and prediction strategies.  
- **Game theory introduction**: Used in psychology and computer science to teach decision-making.

> In short, **Rock, Paper, Scissors** is more than a game — it’s a fast, fair, and surprisingly deep decision-making tool.

## Learning Objectives

- Understand the basic rules and outcomes of Rock, Paper, Scissors.  
- Use RPS as a fair decision-making method.  
- Recognize the game's value in teaching randomness and simple strategies.  
- Implement or play a basic RPS game (with another person or in code).
