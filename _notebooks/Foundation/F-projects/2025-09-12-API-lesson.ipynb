## **Introduction to API Applications**
### What is an API?

An **API (Application Programming Interface)** is a way for different software systems to communicate with each other.  
It defines a set of rules and methods that one program can use to request and exchange data or functionality from another.  

### Why is it useful?

- **Simplifies integration**: Developers don’t need to understand the internal workings of a system — they just use its API.  
- **Saves time**: Instead of rebuilding features, apps can reuse existing services through APIs.  
- **Flexibility**: APIs let you connect different platforms, tools, or databases seamlessly.  
- **Scalability**: They allow software to grow by plugging in new services without rewriting everything.  

In short, APIs are the "bridges" that let software talk to each other efficiently.

We’ll use a test API called JSONPlaceholder. It gives fake user data:

```python
import requests
url = "https://jsonplaceholder.typicode.com/users/1"
response = requests.get(url)

print("Status Code:", response.status_code)
print("Response JSON:", response.json())
```

Most APIs return data in JSON (JavaScript Object Notation). Python can parse it into a dictionary.

```python
import requests
url = "https://jsonplaceholder.typicode.com/users/1"
response = requests.get(url)
data = response.json()
print("Name:", data["name"])
print("Email:", data["email"])
print("Company:", data["company"]["name"])
```

Sometimes you send data to an API (like posting a new blog entry). That’s done with **POST**.

```python
url = "https://jsonplaceholder.typicode.com/posts"
payload = {"title": "Hello API", "body": "Learning APIs!", "userId": 1}

response = requests.post(url, json=payload)
print(response.json())
```

We’ll use a fun API that gives **random cat facts.**

```python
url = "https://catfact.ninja/fact"
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    print("🐱 Cat Fact:", data["fact"])
else:
    print("Error:", response.status_code, response.text)
```

Error Handling is also very important when dealing ith APIs

APIs sometimes fail (bad internet, wrong URL, invalid key). Always check the status code.

```python
import random
import requests
response = requests.get("https://cat-fact.herokuapp.com/facts")

if response.status_code == 200:
    print("Success ✅")
    print(random.choice(response.json())["text"])
else:
    print("Error ❌:", response.status_code)
```
