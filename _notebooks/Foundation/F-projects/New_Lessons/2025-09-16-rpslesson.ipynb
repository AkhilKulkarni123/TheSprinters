# Learning Programming Through Rock Paper Scissors 🎮

## What You'll Learn Today
By looking at this Rock Paper Scissors game, you'll understand the basic building blocks that make ALL computer programs work! Don't worry if this seems complicated at first - every programmer started exactly where you are now.

## 1. Variables: Storing Information 📦

In our game, we need to remember things like:
- What did the player choose? (rock, paper, or scissors)
- What did the computer choose?
- Who won?

**Variables** are like labeled boxes that store information:

```javascript
const playerChoice = "rock";        // Player picked rock
const computerChoice = "scissors";  // Computer picked scissors
let score = 0;                      // Player's current score
```

Think of variables like your phone contacts - you store someone's number under their name so you can find it later!

---

## 2. Functions: Recipes for Actions 🍳

A **function** is like a recipe that the computer can follow over and over. In our game:

```javascript
function playRPS(playerChoice) {
    // 1. Get computer's choice
    // 2. Compare choices  
    // 3. Decide winner
    // 4. Show result
}
```

This is like having a recipe called "Make Sandwich" that you can use whenever you're hungry, just by saying "Make Sandwich with turkey" or "Make Sandwich with ham."

**Why use functions?**
- You write the instructions once, use them many times
- Makes code organized and easier to fix
- Like having a cookbook instead of remembering every recipe

## 3. Objects and Classes: Digital Blueprints 🏗️

Imagine you're building LEGO houses. You have:
- A **blueprint** (shows how to build any house)
- **Actual houses** built from that blueprint

In programming:
- **Class** = Blueprint (instructions for making something)
- **Object** = The actual thing you made

Our game has a `BattleSprite` class (blueprint) that creates rock, paper, and scissors objects:

```javascript
class BattleSprite {
    // Instructions for making any game piece
}

const rockSprite = new BattleSprite();    // Make a rock using the blueprint
const paperSprite = new BattleSprite();   // Make paper using same blueprint
```

**Real-world example:** 
- Car blueprint → Your family's actual car
- Cookie recipe → The cookies you baked yesterday

## 4. Making Decisions: If-Then Logic 🤖

Computers make decisions using **if-then** logic, just like you do:

**You:** "IF it's raining, THEN I'll bring an umbrella"

**Computer:** "IF player chose rock AND computer chose scissors, THEN player wins"

```javascript
if (playerChoice === "rock" && computerChoice === "scissors") {
    result = "You win!";
} else if (playerChoice === computerChoice) {
    result = "Tie!";
} else {
    result = "You lose!";
}
```

This is exactly how the computer decides who wins each round!

---

## 5. Events: Responding to Actions 🖱️

**Events** happen when users interact with the program:
- Clicking a button
- Typing on keyboard
- Moving the mouse

Our game listens for clicks:

```javascript
document.getElementById("rock-btn").addEventListener("click", () => {
    playRPS("rock");  // When rock button clicked, play with rock
});
```

Think of this like a doorbell - when someone presses it (event), the bell rings (your code runs).

---

## 6. Loops and Animation: Making Things Move 🎬

To make smooth animation, the computer needs to:
1. Draw a frame
2. Wait a tiny bit
3. Draw the next frame
4. Repeat forever

This creates the illusion of movement, like a flipbook!

```javascript
function render() {
    // Clear screen
    // Draw background
    // Draw game pieces
    // Wait for next frame
    requestAnimationFrame(render);  // Do this again!
}
```

**Real-world example:** Movies are just 24 pictures per second shown really fast!

---

## 7. How It All Works Together 🔄

Let's trace what happens when you click "Rock":

1. **Event** triggers (you clicked)
2. **Function** `playRPS("rock")` runs
3. **Variable** `playerChoice` stores "rock"
4. **Logic** compares your choice to computer's choice
5. **Function** updates the display
6. **Animation** shows the battle result

It's like a chain reaction where each piece depends on the others!
