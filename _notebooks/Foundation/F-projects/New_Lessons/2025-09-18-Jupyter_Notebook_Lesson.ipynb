# Create & Configure Jupyter Notebooks — Guide

This notebook shows real commands and snippets for:
- creating notebooks
- managing kernels and virtual environments
- installing packages from inside notebooks
- notebook metadata, frontmatter, and filename automation (dates)
- common issues and fixes


## Prerequisites

- Python 3.8+ or Conda installed
- `pip` (or `conda`) available
- `jupyterlab` or `notebook` (we'll show how to install)
- Recommended: use virtual environments or Conda envs to avoid dependency conflicts


```python
# Terminal commands (run in shell)
python -m pip install --upgrade pip
python -m pip install jupyterlab notebook ipykernel jupytext nbconvert
# Optional helpful tools:
python -m pip install black isort pylint         # code formatting / linting
python -m pip install nbformat papermill         # programmatic notebooks
```

Use `%pip` and `%conda` magics in code cells to install packages for the kernel you're running.
This ensures the install targets the kernel environment (safer than system `pip`).


```python
# Inside a notebook cell (works in modern IPython/Jupyter)
%pip install requests rich
# If you're using conda and the conda magics are enabled:
# %conda install -y pandas
```

# If you create a virtualenv or venv, register it as a Jupyter kernel so it appears in the kernel picker.


```python
python -m venv .venv
source .venv/bin/activate        # Windows: .venv\Scripts\activate
pip install --upgrade pip
pip install ipykernel jupyterlab
python -m ipykernel install --user --name myenv --display-name "Python (myenv)"
```

You can programmatically create a new notebook file (.ipynb) or use jupyter-lab/new file UI.
Example: make a basic notebook skeleton with the `nbformat` Python API.


```python
import nbformat
from nbformat.v4 import new_notebook, new_markdown_cell, new_code_cell

nb = new_notebook(cells=[
    new_markdown_cell("# My New Notebook\n\nCreated by script."),
    new_code_cell("print('Hello from new notebook')")
])
with open("starter_notebook.ipynb", "w", encoding="utf-8") as f:
    nbformat.write(nb, f)
print("starter_notebook.ipynb created")
```

# Most importantly

Always remember to use DATES and desscriptive titles when naming notebooks:

-2025-09-18-TITLE.ipynb

Make sure it ends in .ipynb!
