document.getElementById('quizForm').addEventListener('submit', function(e) {
e.preventDefault();
const formData = new FormData(e.target);
quizData = {
sports: Array.from(document.querySelectorAll('input[name="sports"]:checked')).map(el => el.value),
travel: formData.get('travel'),
food: formData.get('food'),
priorities: {
sports: parseInt(formData.get('priority-sports')),

const priorities = [

const uniquePriorities = new Set(priorities);
if (uniquePriorities.size !== 3) {
alert('Please assign unique rankings (1, 2, and 3) to each priority!');

function generateItinerary() {
const prompt = document.getElementById('userPrompt').value.trim();
if (!prompt) {
alert('Please enter a prompt to generate your itinerary!');
return;
}
updateProgress(75);
setTimeout(() => {
const itinerary = createPersonalizedItinerary(prompt);
displayItinerary(itinerary);
showSection('itinerarySection');
updateProgress(100);
}, 800);
}
function createPersonalizedItinerary(prompt) {
const promptLower = prompt.toLowerCase();
let cities = [];
let days = 5;
const daysMatch = prompt.match(/(\d+)[\s-]day/);
if (daysMatch) days = parseInt(daysMatch[1]);
Object.keys(cityData).forEach(city => {
if (promptLower.includes(city.toLowerCase())) cities.push(city);
});
if (cities.length === 0) {
if (promptLower.includes('california') || promptLower.includes('socal')) {
cities = ['San Diego', 'Los Angeles', 'San Francisco'];
} else if (promptLower.includes('pacific northwest') || promptLower.includes('pnw') || promptLower.includes('seattle') || promptLower.includes('portland')) {
cities = ['Seattle', 'Portland'];
} else {
cities = ['San Diego', 'Los Angeles'];
}
}
cities = cities.slice(0, Math.min(cities.length, days));
const itinerary = { title: `Your ${days}-Day West Coast Adventure`, days: [] };
const daysPerCity = Math.max(1, Math.floor(days / cities.length));
let currentDay = 1;
cities.forEach((city, index) => {
const cityDays = (index === cities.length - 1) ? (days - currentDay + 1) : daysPerCity;
const activities = selectActivities(city);
const dayRange = cityDays === 1 ? `Day ${currentDay}` : `Days ${currentDay}-${currentDay + cityDays - 1}`;
itinerary.days.push({ range: dayRange, city: city, activities: activities });
currentDay += cityDays;
});
return itinerary;
}
function selectActivities(city) {
const activities = [];
const cityInfo = cityData[city];
const priorities = [
{ type: 'sports', rank: quizData.priorities.sports },
{ type: 'sightseeing', rank: quizData.priorities.sightseeing },
{ type: 'food', rank: quizData.priorities.food }
];
priorities.sort((a, b) => a.rank - b.rank);
const topPriority = priorities[0].type;
const secondPriority = priorities[1].type;
if (topPriority === 'sports' && cityInfo.sports.length > 0 && !quizData.sports.includes('None')) {
const sportMatch = cityInfo.sports.find(sport => {
return quizData.sports.some(userSport => sport.includes(userSport));
}) || cityInfo.sports[0];
activities.push({ type: 'Sports', description: sportMatch, module: 'sports' });
}
if ((topPriority === 'sightseeing' || secondPriority === 'sightseeing') && cityInfo.destinations.length > 0) {
const count = activities.length === 0 ? 2 : 1;
for (let i = 0; i < Math.min(count, cityInfo.destinations.length); i++) {
activities.push({ type: 'Destination', description: cityInfo.destinations[i], module: 'travel' });
}
}
if ((topPriority === 'food' || secondPriority === 'food') && cityInfo.food.length > 0) {
activities.push({ type: 'Food', description: cityInfo.food[0], module: 'food' });
}
if (activities.length === 0) {
activities.push({ type: 'Destination', description: cityInfo.destinations[0], module: 'travel' });
}
return activities.slice(0, 3);
}

});

}
